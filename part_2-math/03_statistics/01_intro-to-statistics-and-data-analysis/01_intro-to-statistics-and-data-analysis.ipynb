# Introduction to Statistics and Data Analysis

## Overview

Statisticians make use of fundamental laws of probability and statistical inference to draw conclusions about scientific systems.

Information is gathered in the form of **samples**, or **collections of observations**.

**Samples** are collected from **populations**, which are **collections of all individuals** or **individual items of a particular type**.

In an **observational study**, data is collected in the field but the **factors** that influence the results **cannot be preselected**.

In **descriptive statistics**, a set of *single-number* statistics give a sense of center of the location of the data, variability in the data, and the general **nature of the distribution of observations** in the sample.

**Statistical inference** uses techniques that allow us to go beyond merely reporting data to **drawing conclusions** (or inferences) about the system.

## Sampling Procedures: Collection of Data

### Simple Random Sampling

Let us assume that only a single population exists in the problem. **Simple random sampling** implies that any particular sample of a specified sample size has the same chance of being selected as any other sample of the same size.

The term **sample size** means the number of elements in the sample.

If the conditions needed for a random sampling are not followed, you will get a **biased sample**.

In certain cases, random sampling is not always appropriate. Often sampling units are not homogeneous and naturally divide themselves into nonoverlapping groups that are homogeneous. In this cases, it's more appropriate to use a procedure called *stratified random sampling*.

## Measures of Location: The Sample Mean and Median

Measures of location are designed to provide the analyst with some quantitative values of where the center, or some other location, of data is located.

Suppose that the observations in a sample are $ x_1, x_2, \ldots ,x_n $. The **sample mean** denoted by $ \bar{x} $, is

$
\displaystyle
\bar{x} = \sum_{i=1}^{n} \frac{x_1 + x_2 + \ldots + x_n}{n}
$

| Note: |
| :---- |
| The mean is simply a numerical average. |


The **sample median** is another important measure whose purpose is to reflect the central tendency of the sample in such a way that is uninfluenced by extreme values or outliers.

Given that the observations in a sample are $ x_1, x_2, \ldots , x_n $, arranged in **increasing order** of magnitude, the sample median is:

$
\displaystyle
\tilde{x} = \begin{cases}
x_{\frac{n+1}{2}}, \text{ if n is odd}, \\
\frac{x_{\frac{n}{2}} + x_{\frac{n}{2} + 1}}{2}, \text{ if n is even}
\end{cases}
$

### Example

Calculate the mean and median of the following data set:
1.7, 2.2, 3.9, 3.11, 14.7

The sample mean and median is:
$
\bar{x} = \frac{1.7 + 2.2 + 3.9 + 3.11 + 14.7}{5} = 5.12 \\
\tilde{x} = x_\frac{5 + 1}{2} = x_3 = 3.9
$


Note how the sample mean is influenced considerably by the presence of 14.7, whereas the meadian give us the true *center* of the data set.


### Example

Consider the following data set of a study containing the measures of the stem weights in grams of the roots of trees treated and not treated with nitrogen:

| No Nitrogen | Nitrogen |
| :---------- | :------- |
| 0.32 | 0.26 |
| 0.53 | 0.43 |
| 0.28 | 0.47 |
| 0.37 | 0.49 |
| 0.47 | 0.52 |
| 0.43 | 0.75 |
| 0.36 | 0.79 |
| 0.42 | 0.86 |
| 0.38 | 0.62 |
| 0.43 | 0.46 |

Let's use Python to calculate the sample mean and median. Note that before calculating the median, we need to sort the values by order of magnitude (increasingly).

```python
no_nitrogen = [0.32, 0.53, 0.28, 0.37, 0.47, 0.43, 0.36, 0.42, 0.38, 0.43]
nitrogen = [0.26, 0.43, 0.47, 0.49, 0.52, 0.75, 0.79, 0.86, 0.62, 0.46]

def sample_mean(data: list[float]) -> float:
    """Calculate the sample mean of a list of numbers."""
    return sum(data) / len(data)

def sample_median(data: list[float]) -> float:
    """Calculate the sample median of a list of numbers."""
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    if n % 2 != 0:
        return sorted_data[mid]
    else:
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2

print("Sample mean of no nitrogen:", round(sample_mean(no_nitrogen), 3))
print("Sample median of no nitrogen:", round(sample_median(no_nitrogen), 3))
print("Sample mean of nitrogen:", round(sample_mean(nitrogen), 3))
print("Sample median of nitrogen:", round(sample_median(nitrogen), 3))
```

The previous cell was using a custom implementation for the definition of the sample mean and median, but Python also provides a statistics library with those functions:

```python
from statistics import mean, median

no_nitrogen = [0.32, 0.53, 0.28, 0.37, 0.47, 0.43, 0.36, 0.42, 0.38, 0.43]
nitrogen = [0.26, 0.43, 0.47, 0.49, 0.52, 0.75, 0.79, 0.86, 0.62, 0.46]

print("Sample mean of no nitrogen:", round(mean(no_nitrogen), 3))
print("Sample median of no nitrogen:", round(median(no_nitrogen), 3))
print("Sample mean of nitrogen:", round(mean(nitrogen), 3))
print("Sample median of nitrogen:", round(median(nitrogen), 3))
```

## Other measures of location

There are alternatives to the sample mean that are designed to produce values that represent compromises between the mean and the median. However, these are rarely used.

The **trimmed mean**, which is computed by *trimming away* a certain percent of both the largest and the smallest set of values.

For example, the 10% trimmed mean is found by eliminating the largest 10% and smallest 10% and computing the average of the remaining values. It is denoted as $ \bar{x}_{tr(10)} $.


### Example

Caculate the 10% trimmed mean for the following data sets and compare with the median and sample mean:

| No Nitrogen | Nitrogen |
| :---------- | :------- |
| 0.32 | 0.26 |
| 0.53 | 0.43 |
| 0.28 | 0.47 |
| 0.37 | 0.49 |
| 0.47 | 0.52 |
| 0.43 | 0.75 |
| 0.36 | 0.79 |
| 0.42 | 0.86 |
| 0.38 | 0.62 |
| 0.43 | 0.46 |

The 10% trimmed mean is the mean of the samples in the data set after having discarded both the largest 10% and the smallest 10%.

Because the sample size is 10, the 10% is 1 and therefore, we need to eliminate both the smallest and the largest value.

Let's calculate it in Python:

```python
no_nitrogen = [0.32, 0.53, 0.28, 0.37, 0.47, 0.43, 0.36, 0.42, 0.38, 0.43]
nitrogen = [0.26, 0.43, 0.47, 0.49, 0.52, 0.75, 0.79, 0.86, 0.62, 0.46]

def sample_trimmed_mean(data: list[float], trim_percent: float) -> float:
    """Calculate the sample trimmed mean of a list of numbers."""
    n = len(data)
    k = int(n * (trim_percent / 100))
    sorted_data = sorted(data)[k:n-k]
    return sum(sorted_data) / len(sorted_data)

print("Sample 10% trimmed mean of no nitrogen:", round(sample_trimmed_mean(no_nitrogen, 10), 6))
print("Sample 10% trimmed mean of nitrogen:", round(sample_trimmed_mean(nitrogen, 10), 6))
```

Now, we can compare with the sample mean and median:

```python
from statistics import mean, median

print("Sample 10% trimmed mean of no nitrogen:", round(sample_trimmed_mean(no_nitrogen, 10), 6))
print("Sample mean of no nitrogen:", round(mean(no_nitrogen), 6))
print("Sample median of no nitrogen:", round(median(no_nitrogen), 6))
print("=" * 80)
print("Sample 10% trimmed mean of nitrogen:", round(sample_trimmed_mean(nitrogen, 10), 6))
print("Sample mean of nitrogen:", round(mean(nitrogen), 6))
print("Sample median of nitrogen:", round(median(nitrogen), 6))
```
